```python
import os
import numpy as np
import nibabel as nib
import matplotlib.pyplot as plt
```

```python
# ============================================================
# KAGGLE DATASET PATHS
# ============================================================

CT_PATH = "/kaggle/input/datasets/waboke/myct-imgs/RibFrac1-image.nii/RibFrac1-image.nii"

LABEL_PATH = "/kaggle/input/datasets/waboke/myct-imgs/RibFrac1-label.nii/RibFrac1-label.nii"

print("CT path:")
print(CT_PATH)

print("\nLabel path:")
print(LABEL_PATH)
```

```python
# ============================================================
# VERIFY DATASET FILES
# ============================================================

if not os.path.isfile(CT_PATH):
    raise FileNotFoundError(f"CT file not found:\n{CT_PATH}")

if not os.path.isfile(LABEL_PATH):
    raise FileNotFoundError(f"Label file not found:\n{LABEL_PATH}")

print("CT file found successfully.")
print("Label file found successfully.")
```

```python
# ============================================================
# LOAD NIFTI FILES
# ============================================================

ct_img = nib.load(CT_PATH)
label_img = nib.load(LABEL_PATH)

print("CT NIfTI loaded successfully.")
print("Label NIfTI loaded successfully.")
```

```python
# ============================================================
# EXTRACT VOLUMETRIC DATA
# ============================================================

ct_volume = ct_img.get_fdata().astype(np.float32)
label_volume = label_img.get_fdata().astype(np.int16)

print("CT data type:", ct_volume.dtype)
print("Label data type:", label_volume.dtype)
```

```python
# ============================================================
# CHECK VOLUME DIMENSIONS
# ============================================================

print("CT volume shape:", ct_volume.shape)
print("Label volume shape:", label_volume.shape)
```

```python
# ============================================================
# CHECK VOXEL SPACING
# ============================================================

ct_spacing = ct_img.header.get_zooms()[:3]
label_spacing = label_img.header.get_zooms()[:3]

print("CT voxel spacing:", ct_spacing)
print("Label voxel spacing:", label_spacing)
```

```python
# ============================================================
# CT INTENSITY STATISTICS
# ============================================================

print("Minimum CT intensity:", np.min(ct_volume))
print("Maximum CT intensity:", np.max(ct_volume))
print("Mean CT intensity:", np.mean(ct_volume))
print("Standard deviation:", np.std(ct_volume))
```

```python
# ============================================================
# CHECK FRACTURE ANNOTATION VALUES
# ============================================================

unique_labels = np.unique(label_volume)

print("Unique annotation values:")
print(unique_labels)
```

```python
# ============================================================
# ANNOTATION STATISTICS
# ============================================================

total_voxels = label_volume.size

background_voxels = np.sum(label_volume == 0)
annotated_voxels = np.sum(label_volume > 0)

print("Total voxels:", total_voxels)
print("Background voxels:", background_voxels)
print("Annotated voxels:", annotated_voxels)

print(
    "Percentage of annotated voxels:",
    (annotated_voxels / total_voxels) * 100,
    "%"
)
```

```python
# ============================================================
# VISUALIZE CENTRAL CT SLICE
# ============================================================

middle_slice = ct_volume.shape[2] // 2

plt.figure(figsize=(8, 8))

plt.imshow(
    ct_volume[:, :, middle_slice],
    cmap="gray"
)

plt.title(f"CT Slice {middle_slice}")
plt.axis("off")

plt.show()
```

```python
# ============================================================
# VISUALIZE CENTRAL ANNOTATION SLICE
# ============================================================

plt.figure(figsize=(8, 8))

plt.imshow(
    label_volume[:, :, middle_slice]
)

plt.title(f"Fracture Annotation - Slice {middle_slice}")
plt.axis("off")

plt.show()
```

```python
# ============================================================
# CT + FRACTURE ANNOTATION OVERLAY
# ============================================================

plt.figure(figsize=(8, 8))

plt.imshow(
    ct_volume[:, :, middle_slice],
    cmap="gray"
)

plt.imshow(
    label_volume[:, :, middle_slice],
    alpha=0.4
)

plt.title(
    f"CT with Fracture Annotation - Slice {middle_slice}"
)

plt.axis("off")

plt.show()
```

```python
# ============================================================
# FINAL DATA VALIDATION
# ============================================================

assert ct_volume.ndim == 3, \
    "CT volume is not three-dimensional."

assert label_volume.ndim == 3, \
    "Label volume is not three-dimensional."

assert ct_volume.shape == label_volume.shape, \
    "CT and label volumes have different dimensions."

assert np.isfinite(ct_volume).all(), \
    "CT volume contains NaN or infinite values."

assert np.isfinite(label_volume).all(), \
    "Label volume contains NaN or infinite values."

print("==============================================")
print("NOTEBOOK 01 - DATA LOADING SUCCESSFUL")
print("==============================================")

print("CT shape:", ct_volume.shape)
print("Label shape:", label_volume.shape)
print("CT spacing:", ct_spacing)
print("Label spacing:", label_spacing)
print("CT dtype:", ct_volume.dtype)
print("Label dtype:", label_volume.dtype)
print("Unique labels:", unique_labels)

print("==============================================")
```
