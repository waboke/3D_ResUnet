```python
# ============================================================
# NOTEBOOK 07
# 3D RESUNET VISUALIZATION
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D
```

```python
# ============================================================
# PATH CONFIGURATION
# ============================================================

EVALUATION_DIR = "/kaggle/input/3d-resunet-evaluation"

X_TEST_PATH = os.path.join(
    EVALUATION_DIR,
    "X_test.npy"
)

Y_SEG_TEST_PATH = os.path.join(
    EVALUATION_DIR,
    "Y_seg_test.npy"
)

Y_CLASS_TEST_PATH = os.path.join(
    EVALUATION_DIR,
    "Y_class_test.npy"
)

SEG_PRED_PATH = os.path.join(
    EVALUATION_DIR,
    "segmentation_predictions.npy"
)

SEG_PRED_BINARY_PATH = os.path.join(
    EVALUATION_DIR,
    "segmentation_predictions_binary.npy"
)

CLASS_PROB_PATH = os.path.join(
    EVALUATION_DIR,
    "classification_probabilities.npy"
)

CLASS_PRED_PATH = os.path.join(
    EVALUATION_DIR,
    "classification_predictions.npy"
)

SEG_RESULTS_PATH = os.path.join(
    EVALUATION_DIR,
    "segmentation_results.csv"
)

CLASS_RESULTS_PATH = os.path.join(
    EVALUATION_DIR,
    "classification_results.csv"
)
```

```python
# ============================================================
# VERIFY FILES
# ============================================================

paths = [
    X_TEST_PATH,
    Y_SEG_TEST_PATH,
    Y_CLASS_TEST_PATH,
    SEG_PRED_PATH,
    SEG_PRED_BINARY_PATH,
    CLASS_PROB_PATH,
    CLASS_PRED_PATH,
    SEG_RESULTS_PATH,
    CLASS_RESULTS_PATH
]

for path in paths:

    if os.path.exists(path):
        print("FOUND:", path)

    else:
        print("NOT FOUND:", path)
```

```python
# ============================================================
# LOAD TEST DATA
# ============================================================

X_test = np.load(
    X_TEST_PATH
)

Y_seg_test = np.load(
    Y_SEG_TEST_PATH
)

Y_class_test = np.load(
    Y_CLASS_TEST_PATH
)

print("X_test shape:")
print(X_test.shape)

print()

print("Y_seg_test shape:")
print(Y_seg_test.shape)

print()

print("Y_class_test shape:")
print(Y_class_test.shape)
```

```python
# ============================================================
# LOAD PREDICTIONS
# ============================================================

seg_predictions = np.load(
    SEG_PRED_PATH
)

seg_predictions_binary = np.load(
    SEG_PRED_BINARY_PATH
)

class_probabilities = np.load(
    CLASS_PROB_PATH
)

class_predictions = np.load(
    CLASS_PRED_PATH
)

print(
    "Segmentation prediction:",
    seg_predictions.shape
)

print(
    "Classification probability:",
    class_probabilities.shape
)

print(
    "Classification prediction:",
    class_predictions.shape
)
```

```python
# ============================================================
# LOAD EVALUATION TABLES
# ============================================================

segmentation_results = pd.read_csv(
    SEG_RESULTS_PATH
)

classification_results = pd.read_csv(
    CLASS_RESULTS_PATH
)

print(
    segmentation_results.head()
)

print()

print(
    classification_results.head()
)
```

```python
# ============================================================
# VALIDATE DATA
# ============================================================

assert X_test.shape[0] == Y_seg_test.shape[0]

assert X_test.shape[0] == Y_class_test.shape[0]

assert X_test.shape[0] == seg_predictions.shape[0]

assert X_test.shape[0] == class_predictions.shape[0]

print(
    "All test arrays have consistent sample counts."
)
```

```python
# ============================================================
# SELECT TEST PATCH
# ============================================================

PATCH_ID = 0

ct_patch = X_test[PATCH_ID, ..., 0]

true_mask = Y_seg_test[
    PATCH_ID, ..., 0
]

pred_mask = seg_predictions_binary[
    PATCH_ID, ..., 0
]

class_true = int(
    Y_class_test[PATCH_ID]
)

class_pred = int(
    class_predictions[PATCH_ID]
)

class_probability = float(
    class_probabilities[PATCH_ID]
)

print("Patch:", PATCH_ID)

print(
    "True classification:",
    class_true
)

print(
    "Predicted classification:",
    class_pred
)

print(
    "Fracture probability:",
    class_probability
)
```

```python
# ============================================================
# FIND MOST RELEVANT SLICE
# ============================================================

slice_scores = np.sum(
    true_mask,
    axis=(0, 1)
)

best_slice = np.argmax(
    slice_scores
)

print(
    "Best fracture slice:",
    best_slice
)

print(
    "Fracture pixels:",
    slice_scores[best_slice]
)
```

```python
# ============================================================
# DISPLAY CT SLICE
# ============================================================

plt.figure(
    figsize=(6, 6)
)

plt.imshow(
    ct_patch[:, :, best_slice],
    cmap="gray"
)

plt.title(
    f"CT Slice - Patch {PATCH_ID}"
)

plt.axis("off")

plt.show()
```

```python
# ============================================================
# GROUND TRUTH
# ============================================================

plt.figure(
    figsize=(6, 6)
)

plt.imshow(
    true_mask[:, :, best_slice],
    cmap="gray"
)

plt.title(
    "Ground-Truth Rib Fracture Mask"
)

plt.axis("off")

plt.show()
```

```python
# ============================================================
# PREDICTED MASK
# ============================================================

plt.figure(
    figsize=(6, 6)
)

plt.imshow(
    pred_mask[:, :, best_slice],
    cmap="gray"
)

plt.title(
    "Predicted Rib Fracture Mask"
)

plt.axis("off")

plt.show()
```

```python
# ============================================================
# CT + GROUND TRUTH OVERLAY
# ============================================================

plt.figure(
    figsize=(7, 7)
)

plt.imshow(
    ct_patch[:, :, best_slice],
    cmap="gray"
)

plt.imshow(
    true_mask[:, :, best_slice],
    alpha=0.45
)

plt.title(
    "CT + Ground-Truth Fracture Overlay"
)

plt.axis("off")

plt.show()
```

```python
# ============================================================
# CT + PREDICTION OVERLAY
# ============================================================

plt.figure(
    figsize=(7, 7)
)

plt.imshow(
    ct_patch[:, :, best_slice],
    cmap="gray"
)

plt.imshow(
    pred_mask[:, :, best_slice],
    alpha=0.45
)

plt.title(
    "CT + Predicted Fracture Overlay"
)

plt.axis("off")

plt.show()
```

```python
# ============================================================
# THREE-PANEL COMPARISON
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(15, 5)
)

axes[0].imshow(
    ct_patch[:, :, best_slice],
    cmap="gray"
)

axes[0].set_title(
    "CT Image"
)

axes[1].imshow(
    true_mask[:, :, best_slice],
    cmap="gray"
)

axes[1].set_title(
    "Ground Truth"
)

axes[2].imshow(
    pred_mask[:, :, best_slice],
    cmap="gray"
)

axes[2].set_title(
    "3D ResUNet Prediction"
)

for ax in axes:
    ax.axis("off")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# FOUR-PANEL VISUALIZATION
# ============================================================

fig, axes = plt.subplots(
    1,
    4,
    figsize=(20, 5)
)

# CT
axes[0].imshow(
    ct_patch[:, :, best_slice],
    cmap="gray"
)

axes[0].set_title(
    "CT"
)

# Ground truth
axes[1].imshow(
    true_mask[:, :, best_slice],
    cmap="gray"
)

axes[1].set_title(
    "Ground Truth"
)

# Prediction
axes[2].imshow(
    pred_mask[:, :, best_slice],
    cmap="gray"
)

axes[2].set_title(
    "Prediction"
)

# Overlay
axes[3].imshow(
    ct_patch[:, :, best_slice],
    cmap="gray"
)

axes[3].imshow(
    pred_mask[:, :, best_slice],
    alpha=0.45
)

axes[3].set_title(
    "Prediction Overlay"
)

for ax in axes:
    ax.axis("off")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# GROUND TRUTH VS PREDICTION
# ============================================================

intersection = (
    true_mask[:, :, best_slice]
    *
    pred_mask[:, :, best_slice]
)

false_positive = (
    pred_mask[:, :, best_slice]
    -
    intersection
)

false_negative = (
    true_mask[:, :, best_slice]
    -
    intersection
)

plt.figure(
    figsize=(7, 7)
)

plt.imshow(
    ct_patch[:, :, best_slice],
    cmap="gray"
)

plt.imshow(
    false_negative,
    alpha=0.45
)

plt.imshow(
    false_positive,
    alpha=0.45
)

plt.title(
    "Prediction Error Visualization"
)

plt.axis("off")

plt.show()
```

```python
# ============================================================
# CLASSIFICATION RESULT
# ============================================================

true_label = (
    "Fracture"
    if class_true == 1
    else "No Fracture"
)

pred_label = (
    "Fracture"
    if class_pred == 1
    else "No Fracture"
)

print(
    "True class:",
    true_label
)

print(
    "Predicted class:",
    pred_label
)

print(
    f"Fracture probability: "
    f"{class_probability:.4f}"
)
```

```python
# ============================================================
# CLASSIFICATION STATUS
# ============================================================

if class_true == class_pred:

    print(
        "Classification result: CORRECT"
    )

else:

    print(
        "Classification result: INCORRECT"
    )
```

```python
# ============================================================
# VISUALIZE MULTIPLE PATCHES
# ============================================================

NUM_PATCHES = min(
    6,
    len(X_test)
)

for patch_id in range(
    NUM_PATCHES
):

    ct = X_test[
        patch_id, ..., 0
    ]

    true = Y_seg_test[
        patch_id, ..., 0
    ]

    pred = seg_predictions_binary[
        patch_id, ..., 0
    ]

    slice_scores = np.sum(
        true,
        axis=(0, 1)
    )

    slice_id = np.argmax(
        slice_scores
    )

    fig, axes = plt.subplots(
        1,
        3,
        figsize=(12, 4)
    )

    axes[0].imshow(
        ct[:, :, slice_id],
        cmap="gray"
    )

    axes[0].set_title(
        f"CT - Patch {patch_id}"
    )

    axes[1].imshow(
        true[:, :, slice_id],
        cmap="gray"
    )

    axes[1].set_title(
        "Ground Truth"
    )

    axes[2].imshow(
        pred[:, :, slice_id],
        cmap="gray"
    )

    axes[2].set_title(
        "Prediction"
    )

    for ax in axes:
        ax.axis("off")

    plt.tight_layout()

    plt.show()
```

```python
# ============================================================
# BEST SEGMENTATION PATCHES
# ============================================================

best_indices = (
    segmentation_results
    .sort_values(
        "Dice",
        ascending=False
    )
    .head(5)
    ["Patch_ID"]
    .values
)

print(
    "Best patches:"
)

print(
    best_indices
)
```

```python
# ============================================================
# VISUALIZE BEST PATCHES
# ============================================================

for patch_id in best_indices:

    patch_id = int(
        patch_id
    )

    ct = X_test[
        patch_id, ..., 0
    ]

    true = Y_seg_test[
        patch_id, ..., 0
    ]

    pred = seg_predictions_binary[
        patch_id, ..., 0
    ]

    dice = segmentation_results.loc[
        segmentation_results["Patch_ID"]
        == patch_id,
        "Dice"
    ].iloc[0]

    slice_id = np.argmax(
        np.sum(true, axis=(0, 1))
    )

    fig, axes = plt.subplots(
        1,
        3,
        figsize=(12, 4)
    )

    axes[0].imshow(
        ct[:, :, slice_id],
        cmap="gray"
    )

    axes[0].set_title(
        f"CT - Patch {patch_id}"
    )

    axes[1].imshow(
        true[:, :, slice_id],
        cmap="gray"
    )

    axes[1].set_title(
        "Ground Truth"
    )

    axes[2].imshow(
        pred[:, :, slice_id],
        cmap="gray"
    )

    axes[2].set_title(
        f"Prediction\nDice={dice:.4f}"
    )

    for ax in axes:
        ax.axis("off")

    plt.tight_layout()

    plt.show()
```

```python
# ============================================================
# WORST SEGMENTATION PATCHES
# ============================================================

worst_indices = (
    segmentation_results
    .sort_values(
        "Dice",
        ascending=True
    )
    .head(5)
    ["Patch_ID"]
    .values
)

print(
    "Worst patches:"
)

print(
    worst_indices
)
```

```python
# ============================================================
# VISUALIZE WORST PATCHES
# ============================================================

for patch_id in worst_indices:

    patch_id = int(
        patch_id
    )

    ct = X_test[
        patch_id, ..., 0
    ]

    true = Y_seg_test[
        patch_id, ..., 0
    ]

    pred = seg_predictions_binary[
        patch_id, ..., 0
    ]

    dice = segmentation_results.loc[
        segmentation_results["Patch_ID"]
        == patch_id,
        "Dice"
    ].iloc[0]

    slice_id = np.argmax(
        np.sum(true, axis=(0, 1))
    )

    fig, axes = plt.subplots(
        1,
        3,
        figsize=(12, 4)
    )

    axes[0].imshow(
        ct[:, :, slice_id],
        cmap="gray"
    )

    axes[0].set_title(
        f"CT - Patch {patch_id}"
    )

    axes[1].imshow(
        true[:, :, slice_id],
        cmap="gray"
    )

    axes[1].set_title(
        "Ground Truth"
    )

    axes[2].imshow(
        pred[:, :, slice_id],
        cmap="gray"
    )

    axes[2].set_title(
        f"Prediction\nDice={dice:.4f}"
    )

    for ax in axes:
        ax.axis("off")

    plt.tight_layout()

    plt.show()
```

```python
# ============================================================
# CLASSIFICATION RESULTS
# ============================================================

true_classes = Y_class_test.astype(
    int
)

pred_classes = class_predictions.astype(
    int
)

correct = (
    true_classes == pred_classes
)

print(
    "Correct predictions:",
    np.sum(correct)
)

print(
    "Incorrect predictions:",
    np.sum(~correct)
)

print(
    "Total:",
    len(correct)
)
```

```python
# ============================================================
# CLASSIFICATION PROBABILITIES
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.hist(
    class_probabilities[
        true_classes == 0
    ],
    bins=20,
    alpha=0.7,
    label="No Fracture"
)

plt.hist(
    class_probabilities[
        true_classes == 1
    ],
    bins=20,
    alpha=0.7,
    label="Fracture"
)

plt.xlabel(
    "Predicted Fracture Probability"
)

plt.ylabel(
    "Number of Patches"
)

plt.title(
    "Classification Probability Distribution"
)

plt.legend()

plt.show()
```

```python
# ============================================================
# CLASSIFICATION ERRORS
# ============================================================

incorrect_indices = np.where(
    ~correct
)[0]

print(
    "Incorrect classification patches:"
)

print(
    incorrect_indices
)
```

```python
# ============================================================
# VISUALIZE CLASSIFICATION ERRORS
# ============================================================

for patch_id in incorrect_indices[:5]:

    patch_id = int(
        patch_id
    )

    ct = X_test[
        patch_id, ..., 0
    ]

    true = Y_seg_test[
        patch_id, ..., 0
    ]

    pred = seg_predictions_binary[
        patch_id, ..., 0
    ]

    slice_id = np.argmax(
        np.sum(true, axis=(0, 1))
    )

    print(
        f"Patch {patch_id}"
    )

    print(
        "True:",
        true_classes[patch_id]
    )

    print(
        "Predicted:",
        pred_classes[patch_id]
    )

    print(
        "Probability:",
        class_probabilities[patch_id]
    )

    fig, axes = plt.subplots(
        1,
        3,
        figsize=(12, 4)
    )

    axes[0].imshow(
        ct[:, :, slice_id],
        cmap="gray"
    )

    axes[0].set_title(
        "CT"
    )

    axes[1].imshow(
        true[:, :, slice_id],
        cmap="gray"
    )

    axes[1].set_title(
        "Ground Truth"
    )

    axes[2].imshow(
        pred[:, :, slice_id],
        cmap="gray"
    )

    axes[2].set_title(
        "Predicted Segmentation"
    )

    for ax in axes:
        ax.axis("off")

    plt.tight_layout()

    plt.show()
```

```python
# ============================================================
# SAVE BEST AND WORST FIGURES
# ============================================================

def save_comparison_figure(
    patch_id,
    filename
):

    ct = X_test[
        patch_id, ..., 0
    ]

    true = Y_seg_test[
        patch_id, ..., 0
    ]

    pred = seg_predictions_binary[
        patch_id, ..., 0
    ]

    slice_id = np.argmax(
        np.sum(true, axis=(0, 1))
    )

    fig, axes = plt.subplots(
        1,
        3,
        figsize=(12, 4)
    )

    axes[0].imshow(
        ct[:, :, slice_id],
        cmap="gray"
    )

    axes[0].set_title(
        "CT"
    )

    axes[1].imshow(
        true[:, :, slice_id],
        cmap="gray"
    )

    axes[1].set_title(
        "Ground Truth"
    )

    axes[2].imshow(
        pred[:, :, slice_id],
        cmap="gray"
    )

    axes[2].set_title(
        "Prediction"
    )

    for ax in axes:
        ax.axis("off")

    plt.tight_layout()

    path = os.path.join(
        FIGURE_DIR,
        filename
    )

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    return path
```

```python
# ============================================================
# SAVE EXAMPLES
# ============================================================

for i, patch_id in enumerate(
    best_indices
):

    save_comparison_figure(
        int(patch_id),
        f"best_result_{i+1}.png"
    )

for i, patch_id in enumerate(
    worst_indices
):

    save_comparison_figure(
        int(patch_id),
        f"worst_result_{i+1}.png"
    )

print(
    "Best and worst visualization figures saved."
)
```
